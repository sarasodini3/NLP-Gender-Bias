```python
import requests
from bs4 import BeautifulSoup
import re
```

```python
page = requests.get('https://hurstathletics.com/sports/womens-ice-hockey/stats/2024-25/colgate-university/boxscore/14169').content
```

```python
def extract_shot_info(game,period):

  soup = BeautifulSoup(page)
  tables = soup.find_all('table',class_ = 'sidearm-table play-by-play')
  table = tables[period-1]
  rows = table.find_all('tr')

  for row in rows[1:]:
    cells = row.find_all('td')
    time = cells[0].text.strip()

    description = cells[1].text.strip()
    if description == '':
      description = cells[5].text.strip()
    event = description.split(' ')[0]

    event_id = 0

    if event == 'Shot' or event == "GOAL":
      #name = re.sub('.*[A-Z]+ ','',description)
      name = re.sub(f'{event} by \w+ ','',description)
      name_array = name.split(',')
      last = name_array[0]
      first = name_array[1]
      first = re.sub('\.','',first)
      # \ takes away re meaning
      first = re.sub(' Assist.*','',first)
      name = last+','+first
      # . = any character  * = any number of characters [A-Z] = capital letter  + = one or more


      goal = False
      assist = 'NA'
      sog = False
      goalKeeper = 'NA'
      if event == 'GOAL':
        goal = True
        sog = True
        if re.search('Assist',description):
          assist = re.sub('.*Assist by ','', description)[:-1] #slicing off the period at the end
      if re.search('saved', description):
        sog = True
        goalKeeper = re.sub('.*saved by ','', description)[:-1]
      print(f"({game},{period},{event_id},'{time}','{name}',{goal},{sog},'{assist}','{goalKeeper}'),")
```

```python
game = 1
def all_shots(game):
  period = 1
  extract_shot_info(game,period)

  period = 2
  extract_shot_info(game,period)
```

```python
game = 1
all_shots(game)
```
