```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

# Shots Per Game

```python
hurst = pd.read_csv("Mer Hockey Shots.csv")
hurst.head()
```

```python
opp = pd.read_csv("Opp Hockey Shots.csv")
opp.head()
```

```python
# Make x-axis integers for each game

x = np.arange(1, 39, 2)
labels = [str(i) for i in x]
```

```python
# Create Plot

sns.lineplot(x = 'game_id', y = 'num_shots', data = hurst, color = 'green', label = 'Mercyhurst Shots')
sns.lineplot(x = 'game_id', y = 'num_shots', data = opp, color = 'red', label = 'Opponent Shots')
# sns.lineplot(x = 'game_id', y = 'y', data = opp, color = 'Color', label = 'Win, Lose, Draw')
plt.title('Shots per Game')
plt.xticks(x, labels)
plt.xlabel('Game ID')
plt.ylabel('Number of Shots')
plt.legend(title = 'Legend')
plt.show()
```

# SOG Visualisation

```python
hurst_sog = pd.read_csv("SOG Hurst Hockey.csv")
hurst_sog.head()
```

```python
opp_sog = pd.read_csv("SOG Opp Hockey.csv")
opp_sog.head()
```

```python
wdl = pd.read_csv("WDL Hockey.csv")
wdl.head()
```

```python
# Make x-axis integers for each game

x = np.arange(1, 39, 2)
labels = [str(i) for i in x]
```

```python
# Create Plot

sns.lineplot(x = 'game_id', y = 'total_sog', data = hurst_sog, color = 'navy', label = 'Mercyhurst SOG')
sns.lineplot(x = 'game_id', y = 'total_sog', data = opp_sog, color = 'grey', label = 'Opponent SOG')
sns.scatterplot(x = wdl["game_id"], y = wdl["y"], c = wdl["colour"], s = 50)
plt.title('SOG per Game')
plt.xticks(x, labels)
plt.xlabel('Game ID')
plt.ylabel('Number of SOG')
plt.legend(title = 'Legend')
plt.show()
```

# Total Shots By Height

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
shots_by_height = pd.read_csv('height_shots.csv')
shots_by_height
```

```python
sns.lineplot(x= 'height', y= 'sum', data= shots_by_height, color='red')
plt.xlabel('Height (in.)')
plt.ylabel('Total Shots')
plt.title('Total Shots by Height')
plt.show()
```

# Penalty Minutes By Class

```python
class_penalties = pd.read_csv('class_penalties.csv')
class_penalties
```

```python
colors = ['#ff6384', '#36a2eb', '#ffcd56', '#4bc0c0', '#9966ff']
plt.pie(class_penalties['sum'], labels= class_penalties['class'], colors = colors)
plt.title('Number of Penalties by Class')

plt.show()
```

# Time first goal is scored

```python
first_goal = pd.read_csv("First Goal.csv")
first_goal.head()
```

```python
# Make x-axis integers for each game

x = np.arange(1, 39, 2)
labels = [str(i) for i in x]
```

```python
sns.lineplot(x = first_goal["game_id"], y = first_goal["min_t"], c = '#d7d7d7', zorder = 1)
sns.scatterplot(x = first_goal["game_id"], y = first_goal["min_t"], c = first_goal["colour"], s = 50, zorder = 2)
plt.title('First Goal Per Game')
plt.xticks(x, labels)
plt.xlabel('Game ID')
plt.ylabel('Time')
plt.show()
```
