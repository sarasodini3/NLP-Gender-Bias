```python
import requests
from bs4 import BeautifulSoup
```

```python
def get_game_stats(game, page):
  soup = BeautifulSoup(page)
  heading = soup.find_all('h4', class_="sub-heading")[1].text[:-2]
  if heading == 'Mercyhurst':
    table = soup.find_all('table')[5]
  else:
    table = soup.find_all('table')[6]

  if table.find_all('tr')[1].text.strip()[-1] == '3':
    overtime = False
  else:
    overtime = True

  for i in range(2,len(table.find_all('tr'))-2):
    tr = table.find_all('tr')[i]
    tds = tr.find_all('td')
    number = tds[1].text.strip()
    if tds[2].find('a'):
      name = tds[2].find('a').text.strip()
    else:
      name = tds[2].text.strip()
      name = name[1:-1]
    goals = tds[3].text.strip()
    assists = tds[4].text.strip()
    pd1_shots = tds[5].text.strip()
    pd2_shots = tds[6].text.strip()
    pd3_shots = tds[7].text.strip()

    if overtime == True:
      pd4_shots = tds[8].text.strip()
      totals = tds[9].text.strip()
      plus_minus = tds[11].text.strip()
      face_off = tds[12].text.strip()
      if face_off == '-':
          face_off_won = '0'
          face_off_lost = '0'
      else:
        face_off_won = face_off.split('-')[0]
        face_off_lost = face_off.split('-')[1]
      penalty = tds[13].text.strip()
      if penalty == '-':
        penalty_amount = '0'
        penalty_mins = '0'
      else:
        penalty_amount = penalty.split('-')[0]
        penalty_mins = penalty.split('-')[1]
      blocked = tds[14].text.strip()
      print(f"({game}, {number}, '{name}', {goals}, {assists}, {pd1_shots}, {pd2_shots}, {pd3_shots}, {pd4_shots}, 'NA', {totals}, {plus_minus}, {face_off_won}, {face_off_lost}, {penalty_amount}, {penalty_mins}, {blocked}),")

    else:
      totals = tds[8].text.strip()
      plus_minus = tds[10].text.strip()
      face_off = tds[11].text.strip()
      if face_off == '-':
          face_off_won = '0'
          face_off_lost = '0'
      else:
        face_off_won = face_off.split('-')[0]
        face_off_lost = face_off.split('-')[1]
      penalty = tds[12].text.strip()
      if penalty == '-':
        penalty_amount = '0'
        penalty_mins = '0'
      else:
        penalty_amount = penalty.split('-')[0]
        penalty_mins = penalty.split('-')[1]
      blocked = tds[13].text.strip()
      print(f"({game}, {number}, '{name}', {goals}, {assists}, {pd1_shots}, {pd2_shots}, {pd3_shots}, 'NA', 'NA', {totals}, {plus_minus}, {face_off_won}, {face_off_lost}, {penalty_amount}, {penalty_mins}, {blocked}),")
```

```python
schedule = requests.get('https://hurstathletics.com/sports/womens-ice-hockey/schedule').content
```

```python
soup = BeautifulSoup(schedule)
items = soup.find_all('li', class_='sidearm-schedule-game-links-boxscore')
urls=[]
for i in range(0,76,2):
  url = 'https://hurstathletics.com'+items[i].find('a')['href']
  urls.append(url)
```

```python
urls
```

```python
game = 1
for url in urls:
  page = requests.get(url).content
  if game == 7 or game == 30:
    get_game_stats_shutout(game, page)
  else:
    get_game_stats(game, page)
  game += 1
```
